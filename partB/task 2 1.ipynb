# Task 2.1 Dataset Selection and Setup

## Dataset Choice
For this reproduction, I am using a synthetic binary classification dataset generated using `scikit-learn`'s `make_classification`. 

**Justification:**
- **What it is:** A 2D classification dataset with 100 samples and 2 informative features, featuring a small amount of class overlap (noise).
- **Why it is a reasonable testbed:** The paper's core contribution is the efficient computation of the SVM regularization path. A 2D toy dataset allows for clear visualization of the support vectors and the decision boundary as the regularization parameter $c$ (or $C$) varies. It provides a simple yet effective environment to verify that the suboptimal path correctly approximates the exact solution.
- **Limitations:** Compared to the original paper (which used datasets like 'spam' or 'ad' with hundreds or thousands of features), this dataset is extremely low-dimensional. This means the computational bottleneck of large inverse matrix updates is less pronounced, but the algorithm's logic remains identical.

## Dataset Preparation

```python
import numpy as np
from sklearn.datasets import make_classification
import pandas as pd
import os

# Set random seed for reproducibility
np.random.seed(42)

# Generate toy dataset
X, y = make_classification(n_samples=100, n_features=2, n_informative=2, 
                           n_redundant=0, n_clusters_per_class=1, 
                           flip_y=0.05, class_sep=1.0, random_state=42)

# Change y from {0, 1} to {-1, +1} as per SVM convention in the paper
y = 2 * y - 1

# Save to partB/data/
if not os.path.exists('data'):
    os.makedirs('data')

df = pd.DataFrame(X, columns=['feature1', 'feature2'])
df['target'] = y
df.to_csv('data/toy_dataset.csv', index=False)

# Create README in data/
with open('data/README.md', 'w') as f:
    f.write('# Dataset Information\n\n')
    f.write('The `toy_dataset.csv` contains a synthetic binary classification dataset generated with `make_classification`.\n')
    f.write('- Samples: 100\n')
    f.write('- Features: 2 (feature1, feature2)\n')
    f.write('- Labels: -1, +1\n')

print("Dataset generated and saved to partB/data/")
df.head()
```
