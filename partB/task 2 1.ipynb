# Task 2.1: Dataset Selection and Setup

### Choice of Dataset
I have chosen a synthetic time-series dataset representing **Sine Waves vs. Square Waves**. 

**Justification (3-5 sentences):**
This dataset is a reasonable testbed because shapelets are specifically designed to capture 'local' shapes like the smooth curvature of a sine wave versus the abrupt edges and flat tops of a square wave. It provides a clear, interpretable discriminative feature that directly tests the core contribution of the Shapelet Transform. While it contains at least 150 samples and 50 features (fulfilling the requirement), it is limited compared to the original paper's datasets because it lacks the multi-class complexity and high noise levels found in real-world UCR datasets. 

### Preprocessing
- **Z-Normalization:** Each time series is normalized to have a mean of 0 and a standard deviation of 1. This follows the paper's recommendation (Section 2.1) to ensure the distance metric is based on shape rather than amplitude or offset.

```python
import numpy as np
import matplotlib.pyplot as plt
import os
np.random.seed(42)

def generate_toy_data(n_samples=150, n_steps=50):
    X = []; y = []
    t = np.linspace(0, 4*np.pi, n_steps)
    
    for _ in range(n_samples // 2):
        # Class 0: Sine wave + noise
        series_0 = np.sin(t) + np.random.normal(0, 0.2, n_steps)
        X.append((series_0 - np.mean(series_0)) / np.std(series_0))
        y.append(0)
        
        # Class 1: Square wave + noise
        series_1 = np.sign(np.sin(t)) + np.random.normal(0, 0.2, n_steps)
        X.append((series_1 - np.mean(series_1)) / np.std(series_1))
        y.append(1)
        
    return np.array(X), np.array(y)

X, y = generate_toy_data()
plt.figure(figsize=(10, 4))
plt.plot(X[0], label='Normalized Sine')
plt.plot(X[1], label='Normalized Square')
plt.title("Toy Dataset: Sine vs Square (Z-Normalized)")
plt.legend()
os.makedirs('results', exist_ok=True)
plt.savefig('results/toy_data_samples.png')
plt.show()
```
