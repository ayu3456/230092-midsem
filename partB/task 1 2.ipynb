# Task 1.2 Key Assumptions

### Assumption 1
**Assumption:** The matrix $M = \begin{pmatrix} 0 & y_M^\top \\ y_M & Q_M \end{pmatrix}$ is non-singular during the path segment.

**Why the method needs it:** This matrix must be invertible to solve the linear system in Theorem 1 (Equation 5) and obtain the update gradients $\beta$ and $g$. If it is singular, the path direction is undefined.

**Violation scenario:** This assumption is violated if the sub-matrix $Q_M$ is positive semi-definite but not positive definite on the relevant subspace, which can happen with certain kernels (like a linear kernel) on highly redundant or perfectly collinear data points.

**Paper reference:** Theorem 1, Footnote 2 in Section 2.2.

### Assumption 2
**Assumption:** The solution path is piecewise-linear with respect to the regularization parameter $c$.

**Why the method needs it:** The algorithm relies on the fact that as long as the index sets $\pi = (O, M, I)$ don't change, the optimal $\alpha$ changes linearly with $c$. The "step length" calculation for breakpoints only works for linear segments.

**Violation scenario:** This assumption would be violated if the objective function were not a Quadratic Program (QP) or if the regularization term were not an $L_2$ norm (e.g., if using a different loss function that makes the dual non-quadratic).

**Paper reference:** Section 1 (Introduction), Paragraph 2: "In a class of parametric quadratic programs (QPs), the solution path is represented as a piecewise-linear function".

### Assumption 3
**Assumption:** The "optimal" solution can be approximated within a fixed tolerance $\epsilon_1, \epsilon_2$ without significant loss in generalization performance.

**Why the method needs it:** The entire computational efficiency gain comes from the "suboptimality." If the user requires perfect convergence to the exact KKT conditions, the number of breakpoints defaults back to many redundant ones, negating the algorithm's benefit.

**Violation scenario:** In sensitive applications where highly precise margins are required for stability (e.g., certain theoretical physics simulations or high-stakes financial arbitrage where small boundary shifts matter), this approximation might lead to unstable predictions.

**Paper reference:** Section 1 (Introduction): "the strict optimality... adversely affects the computational efficiency... generalization performances of these suboptimal solutions are not significantly different".
