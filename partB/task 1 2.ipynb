# Task 1.2: Key Assumptions

### Assumption 1
**Assumption:** Local discriminability -- class differences are captured by short, local subsequences rather than global statistical trends.
**Why the method needs it:** The entire transform relies on finding 'shapelets' (short sequences). If the class difference is purely frequency-based or global (e.g., total energy), a local match won't distinguish classes.
**Violation scenario:** Differentiating two sine waves that only differ in their total duration or overall frequency, where any local 10-point window looks identical.
**Paper reference:** Section 1 (Introduction) and Section 2.1 (Definition of Shapelet).

### Assumption 2
**Assumption:** Phase-independence -- the discriminative shapelet can occur at any index within the time series.
**Why the method needs it:** The $dist(T, S)$ calculation uses a 'sliding window' minimum. This assumes the feature's location is irrelevant to its meaning.
**Violation scenario:** An ECG signal where a specific spike at the *start* of the cycle means one condition, but a spike at the *end* means another.
**Paper reference:** Section 2.1 (Subsequence Distance definition).

### Assumption 3
**Assumption:** Euclidean distance sufficiency -- once segments are Z-normalized, Euclidean distance is an adequate measure of shape similarity.
**Why the method needs it:** The search space for shapelets is already high ($O(n^4)$); using complex metrics like Dynamic Time Warping (DTW) for every candidate would be computationally infeasible.
**Violation scenario:** Data with significant elastic warping where peaks do not align in time, making Euclidean distance high even for identical shapes.
**Paper reference:** Section 2.2 (Distance Measures).
