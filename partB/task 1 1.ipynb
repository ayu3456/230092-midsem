**Paper:** A shapelet transform for time series classification (Lines et al., 2012)

### Step-by-Step Method Description

The Shapelet Transform (ST) decouples the discovery of shapelets from the classifier. The process operates as follows:

**Step 1: Shapelet Candidate Extraction**
- **Description:** Generate all possible subsequences of lengths in a range $[min, max]$ from the training set.
- **Reference:** Section 3.1 (Algorithm 1: `GenerateCandidates`).
- **Purpose:** To populate a candidate pool from which highly discriminative subsequences can be selected.

**Step 2: Subsequence Distance Calculation**
- **Description:** For each candidate shapelet $S$, calculate the minimum Euclidean distance between $S$ and every time series $T$ in the training set: $dist(T, S) = \min_{T'} d(T', S)$ where $T'$ is a subsequence of $T$ of $|S|$ length.
- **Reference:** Section 2.1 (Subsequence Distance).
- **Purpose:** To transform a variable-length time series into a single numeric distance feature relative to a specific shape.

**Step 3: Quality Assessment (Information Gain)**
- **Description:** Evaluate the quality of each candidate by sorting the training series by their distance to the shapelet and finding the optimal split point that maximizes Information Gain.
- **Reference:** Section 3.2 (Shapelet Quality Measures).
- **Purpose:** To rank candidates by their ability to distinguish between classes.

**Step 4: Self-Similar Shapelet Filtering**
- **Description:** Identify candidates that are slightly shifted versions of the same original sequence and retain only the highest-quality one to avoid redundancy.
- **Reference:** Section 3.3 (Removing Similar Shapelets).
- **Purpose:** To ensure the final feature set is diverse and not dominated by nearly identical subsequences.

**Step 5: Dimensionality Mapping (Transform)**
- **Description:** Select the top-$k$ shapelets and transform the original dataset $D$ into a new dataset $D'$ where each instance is a vector $[dist(T, S_1), dist(T, S_2), ..., dist(T, S_k)]$.
- **Reference:** Section 3.4 (The Shapelet Transform).
- **Purpose:** To create a standard attribute-value representation compatible with any machine learning classifier.

**Step 6: High-Performance Classification**
- **Description:** Apply a robust classifier, such as a Support Vector Machine or Rotation Forest, to the transformed $k$-dimensional feature space.
- **Reference:** Section 4 (Experimental Results).
- **Purpose:** To achieve superior accuracy by leveraging global classifier optimization on the extracted local features.

### Final Summary Sentence
This paper solves the limitation of the original Shapelet Tree's instability and inability to use ensemble methods by transforming time series into a distance-based feature space, which the authors claim enables more accurate classification while retaining the interpretability of local motifs.
