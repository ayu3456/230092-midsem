# Task 1.3: What the Paper Claims to Improve

**1. Baseline Method:** 
The primary prior method is the **Shapelet Tree** (Ye and Keogh, 2009).

**2. Limitation Identified:** 
The authors argue that the Shapelet Tree is unstable and suffers from the 'greedy' nature of decision trees. Furthermore, it limits the user to a single classifier type (Decision Tree), which often underperforms compared to ensembles or kernel-based methods on complex tasks.

**3. Proposed Improvement:** 
The paper introduces a **Transform** that extracts shapelets as features *before* classification. This 'decoupling' allows the use of high-performance classifiers like Rotation Forest or Support Vector Machines, which can consider all shapelets simultaneously rather than hierarchically.

**4. Failure Scenario vs. Baseline:** 
The proposed Shapelet Transform might not outperform the Shapelet Tree on extremely large datasets where the memory overhead of storing the full transformed distance matrix ($N$ samples $\times K$ shapelets) becomes a bottleneck compared to the memory-efficient path traversal of a decision tree.
